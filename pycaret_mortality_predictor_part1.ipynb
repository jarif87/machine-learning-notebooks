```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
pd.set_option('display.max_columns',None)
%matplotlib inline
```

```python
df=pd.read_csv('/content/mortality.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop(columns=["ID"],axis=1,inplace=True)
```

```python
df.head()
```

```python
plt.rcParams.update({'figure.figsize': (15, 6), 'axes.titlesize': 16, 'axes.labelsize': 12})
sns.countplot(x='outcome', data=df)
plt.title('Count of Outcomes')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.show()
```

```python
sns.histplot(df['age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(x='outcome', y='BMI', data=df)
plt.title('BMI by Outcome')
plt.xlabel('Outcome')
plt.ylabel('BMI')
plt.show()
```

```python
sns.scatterplot(x='heart rate', y='Systolic blood pressure', hue='outcome', size='BMI', data=df)
plt.title('Heart Rate vs Systolic Blood Pressure')
plt.xlabel('Heart Rate')
plt.ylabel('Systolic Blood Pressure')
plt.show()
```

```python
corr = df.corr()
sns.heatmap(corr, cmap='RdYlGn', annot=False, center=0)
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.violinplot(x='gendera', y='Creatinine', hue='outcome', split=True, data=df)
plt.title('Creatinine Distribution by Gender and Outcome')
plt.xlabel('Gender (1: Male, 2: Female)')
plt.ylabel('Creatinine')
plt.show()
```

```python
ts = df['heart rate'].cumsum()
ts.plot(c='r')
plt.title('Cumulative Heart Rate Over Time')
plt.xlabel('Index')
plt.ylabel('Cumulative Heart Rate')
plt.show()
```

```python
hypertensive_counts = df['hypertensive'].value_counts()
plt.pie(hypertensive_counts, labels=['Non-Hypertensive', 'Hypertensive'], autopct='%1.1f%%', startangle=90)
plt.title('Proportion of Hypertensive Patients')
plt.show()
```

```python
df[['Blood potassium', 'Blood sodium', 'Blood calcium']].plot.area(stacked=False)
plt.title('Blood Parameters Over Observations')
plt.xlabel('Index')
plt.ylabel('Value')
plt.show()
```

```python
plt.hexbin(df['heart rate'], df['Respiratory rate'], gridsize=20, cmap='Blues')
plt.colorbar(label='Count')
plt.title('Hexagonal Bin Plot: Heart Rate vs Respiratory Rate')
plt.xlabel('Heart Rate')
plt.ylabel('Respiratory Rate')
plt.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.palettes import magma
from pandas.plotting import scatter_matrix, andrews_curves, autocorrelation_plot
```

```python
andrews_curves(df[['age', 'BMI', 'heart rate', 'Systolic blood pressure', 'outcome']], 'outcome')
plt.title('Andrews Curves by Outcome')
plt.show()
```

```python
fig = px.scatter(df, x='heart rate', y='Systolic blood pressure', color='outcome', size='BMI',
                 hover_data=['age', 'gendera'], title='Interactive Scatter: Heart Rate vs Systolic BP')
fig.show()
```

```python
comorbidities = ['hypertensive', 'atrialfibrillation', 'CHD with no MI', 'diabetes', 'deficiencyanemias', 'depression', 'Hyperlipemia', 'Renal failure', 'COPD']
comorbidity_counts = df.groupby('outcome')[comorbidities].sum()
comorbidity_counts.plot(kind='bar', stacked=True)
plt.title('Comorbidities by Outcome')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.show()
```

```python
for outcome in df['outcome'].unique():
    sns.kdeplot(df[df['outcome'] == outcome]['BMI'], label=f'Outcome {outcome}')
plt.title('BMI Density by Outcome')
plt.xlabel('BMI')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
autocorrelation_plot(df['heart rate'])
plt.title('Autocorrelation of Heart Rate')
plt.show()
```

```python
fig = px.scatter_3d(df, x='heart rate', y='Systolic blood pressure', z='BMI', color='outcome',
                    title='3D Scatter: Heart Rate, Systolic BP, BMI')
fig.show()
```

```python
g = sns.FacetGrid(df, col='gendera', row='outcome', margin_titles=True)
g.map(sns.histplot, 'Systolic blood pressure')
g.set_axis_labels('Systolic Blood Pressure', 'Count')
g.fig.suptitle('Systolic BP by Gender and Outcome', y=1.02)
plt.show()
```

```python
sns.boxenplot(x='outcome', y='glucose', data=df)
plt.title('Glucose Distribution by Outcome')
plt.xlabel('Outcome')
plt.ylabel('Glucose')
plt.show()
```
