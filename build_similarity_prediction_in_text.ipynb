```python
from sentence_transformers import SentenceTransformer, util
import numpy as np
```

```python
# Load the pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
text1="J. Walk (@Jwalk35_) commented 18 hours ago, sharing his thoughts on a recent article he read. In his view, the writer made a compelling argument that, rather than reshaping existing male characters like James Bond to fit female leads, creators should instead focus on crafting original, powerful, and distinctive female protagonists—characters who can stand on their own and potentially even surpass the cultural legacy of figures like James Bond."
text2="Joe Lara, best known for his lead role in the popular 1990s adventure television series Tarzan: The Epic Adventures, has tragically died in a plane crash. The 58-year-old actor was traveling with his wife and several others when the small private aircraft went down. Fans around the world are mourning the loss of a performer who once embodied one of fiction’s most iconic jungle heroes."
```

```python
embeddings1 = model.encode(text1, convert_to_tensor=True)
embeddings2 = model.encode(text2, convert_to_tensor=True)
```

```python
# Compute cosine similarity
similarity = util.cos_sim(embeddings1, embeddings2)
score = similarity.item()

# Create a user-friendly explanation
if score >= 0.7:
    explanation = "These sentences are very similar in meaning."
elif score >= 0.3:
    explanation = "These sentences are somewhat related but not very similar."
else:
    explanation = "These sentences are mostly unrelated in meaning."

# Print the score and explanation
print(f"Similarity Score: {score:.4f}")
print(explanation)
```
