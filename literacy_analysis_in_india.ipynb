```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
pd.options.display.max_columns=None
%matplotlib inline
```

```python
plt.figure(figsize=(10, 12))
sns.barplot(x='TOTPOPULAT', y='STATNAME', data=df,palette="Dark2")
plt.xticks(rotation=90)
plt.xlabel('Population', fontsize=14)
plt.ylabel('States in India', fontsize=14)
plt.title('Population with respect to States in India', fontsize=16)
plt.show()
```

```python
cols=['STATNAME','OVERALL_LI','FEMALE_LIT','MALE_LIT','SEXRATIO','DISTRICTS','TOTPOPULAT','P_URB_POP','AREA_SQKM','SCHTOT','SCHTOTG','SCHTOTGR','SCHTOTPR','ENRTOT','ENRTOTG','ENRTOTGR',
       'ENRTOTPR', 'TCHTOTG', 'TCHTOTP', 'SCLSTOT', 'STCHTOT', 'ROADTOT', 'SPLAYTOT', 'SWATTOT',  'SELETOT']
df=df[cols]
```

```python
df.head()
```

```python
df['District_Label'] = [f'District {i+1}' for i in range(len(df))]
```

```python
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (20, 4)

df_subset = df.head(100)

bar_width = 1
index = np.arange(len(df_subset))

plt.bar(index, df_subset['OVERALL_LI'], bar_width, label='Overall Literacy', color='skyblue')
plt.bar(index + bar_width, df_subset['FEMALE_LIT'], bar_width, label='Female Literacy', color='pink')
plt.bar(index + 2 * bar_width, df_subset['MALE_LIT'], bar_width, label='Male Literacy', color='lightgreen')

plt.xlabel('District')
plt.ylabel('Literacy Rate (%)')
plt.title('Literacy Rates by District (First 100)')
plt.xticks(index + bar_width, df_subset['District_Label'], rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
df_subset = df.head(100)
index = np.arange(len(df_subset))

bar_width = 1

plt.bar(index, df_subset['SCHTOT'], bar_width, label='Total Schools', color='skyblue')
plt.bar(index + bar_width, df_subset['SCHTOTG'], bar_width, label='Government Schools', color='orange')
plt.bar(index + 2 * bar_width, df_subset['SCHTOTPR'], bar_width, label='Private Rural Schools', color='lightgreen')

plt.xlabel('District')
plt.ylabel('Number of Schools')
plt.title('School Counts by District (First 100)')
plt.xticks(index + bar_width, df_subset['District_Label'], rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
fig = px.scatter(df_subset, x='P_URB_POP', y='OVERALL_LI', text='District_Label', size='TOTPOPULAT',
                 color='OVERALL_LI', color_continuous_scale='Viridis',
                 title='Overall Literacy vs. Urban Population Percentage',
                 labels={'P_URB_POP': 'Urban Population (%)', 'OVERALL_LI': 'Overall Literacy (%)'})

fig.update_traces(textposition='top center')
fig.show()
```

```python
plt.figure(figsize=(12, 8))
corr = df[['OVERALL_LI', 'FEMALE_LIT', 'MALE_LIT', 'SEXRATIO', 'TOTPOPULAT',
           'P_URB_POP', 'AREA_SQKM', 'SCHTOT', 'ENRTOT', 'TCHTOTG', 'TCHTOTP']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of Key Metrics')
plt.tight_layout()
plt.show()
```

```python

sns.histplot(df['TOTPOPULAT'], bins=4, kde=True, color='purple')
plt.xlabel('Total Population')
plt.ylabel('Frequency')
plt.title('Population Distribution Across Districts')
plt.tight_layout()
plt.show()
```

```python
plt.bar(df_subset['District_Label'], df_subset['ENRTOTG'], label='Government Enrollment', color='skyblue')
plt.bar(df_subset['District_Label'], df_subset['ENRTOTPR'], bottom=df_subset['ENRTOTG'], label='Private Rural Enrollment', color='orange')
plt.xlabel('District')
plt.ylabel('Enrollment')
plt.title('Enrollment Breakdown by District')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
infrastructure = df[['ROADTOT', 'SPLAYTOT', 'SWATTOT', 'SELETOT']].melt()
sns.boxplot(x='variable', y='value', data=infrastructure, palette='Set3')
plt.xlabel('Infrastructure Type')
plt.ylabel('Number of Schools')
plt.title('Distribution of School Infrastructure')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
district = df.iloc[0]
plt.figure(figsize=(8, 8))
plt.pie([district['SCHTOTG'], district['SCHTOTPR']],
        labels=['Government Schools', 'Private Rural Schools'],
        colors=['skyblue', 'lightgreen'], autopct='%1.1f%%')
plt.title(f'School Types in {district["District_Label"]}')
plt.tight_layout()
plt.show()
```

```python
df_subset['Student_Teacher_Ratio'] = df_subset['ENRTOT'] / (df_subset['TCHTOTG'] + df_subset['TCHTOTP'])
sns.barplot(x='District_Label', y='Student_Teacher_Ratio', data=df_subset, palette='Blues')
plt.xlabel('District')
plt.ylabel('Student-Teacher Ratio')
plt.title('Student-Teacher Ratio by District')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig = px.parallel_coordinates(
    df,
    dimensions=['OVERALL_LI', 'FEMALE_LIT', 'MALE_LIT', 'P_URB_POP', 'SCHTOT'],
    color='OVERALL_LI',
    color_continuous_scale=px.colors.sequential.Inferno,
    title='Parallel Coordinates Plot of Key Metrics',
    labels={'District_Label': 'District'}
)
fig.show()
```

```python
top_bottom = pd.concat([
    df.nlargest(5, 'OVERALL_LI'),  # Top 5 states by overall literacy
    df.nsmallest(5, 'OVERALL_LI')  # Bottom 5 states by overall literacy
], axis=0)
```

```python
sns.barplot(x='STATNAME', y='OVERALL_LI', data=top_bottom, alpha=0.8,palette="bwr")
plt.xticks(rotation=45)
plt.ylabel('Literacy Rate (%)', fontsize=14)
plt.xlabel('States in India', fontsize=14)
plt.title('Top 5 and Bottom 5 States by Overall Literacy Rate', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
top_bottom['Difference'] = top_bottom['MALE_LIT'] - top_bottom['FEMALE_LIT']
sns.barplot(x='STATNAME', y='Difference', data=top_bottom, alpha=0.8,palette="Set1")
plt.xticks(rotation=45)
plt.ylabel('Literacy Rate Difference (%)', fontsize=14)
plt.xlabel('States in India', fontsize=14)
plt.title('Top 3 and Bottom 3 States by Literacy Rate Difference', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
top_bottom['Rural']=100-top_bottom['P_URB_POP']
top_bottom['Rural']
```

```python
df_melted = top_bottom.melt(id_vars='STATNAME', value_vars=['P_URB_POP', 'Rural'],
                            var_name='Population_Type', value_name='Population')

sns.barplot(x='STATNAME', y='Population', hue='Population_Type', data=df_melted)
plt.xticks(rotation=45)
plt.xlabel('States', fontsize=14)
plt.ylabel('Population', fontsize=14)
plt.title('Comparison of Rural and Urban Population', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
sns.barplot(x='STATNAME', y='SEXRATIO', data=top_bottom, alpha=0.8,palette="hot")
plt.xticks(rotation=45)
plt.xlabel('States', fontsize=14)
plt.ylabel('Sex Ratio', fontsize=14)
plt.title('Sex Ratio by State', fontsize=16)
plt.tight_layout()
plt.show()
```
